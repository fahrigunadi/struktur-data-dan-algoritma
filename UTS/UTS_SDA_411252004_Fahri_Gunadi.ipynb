# SOAL 1 - Pencarian Data Mahasiswa dan Pengurutan Rekursif

Alur: data disimpan dalam list -> dicari manual dengan Linear Search -> diurutkan manual dengan Merge Sort (rekursif) -> dicari lagi dengan Binary Search manual. Dilarang memakai `sorted()`, `list.sort()`, `bisect`, atau library sort/search untuk bagian manual.

## 1.1 Data awal (d = 4)
Nilai Bima = 80+d, Hadi = 65+d, Lukman = 70+d. Target pasti ditemukan: `target_index = d % len(data)` -> `target_nim`.

```python
# Identitas dan data awal (mengikuti modul Pertemuan 2: array & pencarian)
NIM = "411252004"
d = int(NIM[-1])  # d = 4
print("NIM:", NIM, "-> d =", d)

data = [
    ['TI001', 'Alya', 78],
    ['TI002', 'Bima', 80 + d],
    ['TI003', 'Citra', 67],
    ['TI004', 'Danu', 92],
    ['TI005', 'Eka', 74],
    ['TI006', 'Fajar', 88],
    ['TI007', 'Gina', 81],
    ['TI008', 'Hadi', 65 + d],
    ['TI009', 'Intan', 91],
    ['TI010', 'Joko', 76],
    ['TI011', 'Karin', 84],
    ['TI012', 'Lukman', 70 + d],
]

# Target yang pasti ditemukan + satu target yang tidak ada
target_index = d % len(data)
target_nim = data[target_index][0]
tidak_ada = 'TI999'
print("target_index =", d, "%", len(data), "=", target_index)
print("target_nim =", target_nim, "| kode tidak ada =", tidak_ada)
print("\nData awal:")
for i, row in enumerate(data):
    print(f"  [{i:2d}] {row[0]} {row[1]:6s} {row[2]}")
```

## 1.2 Linear Search manual
Cek satu per satu dari indeks 0. Tiap baris yang dibandingkan dihitung 1 perbandingan. Kembali `(indeks, jumlah_perbandingan)`, `-1` kalau tidak ketemu.

```python
def linear_search_nim(data, target):
    """Cari kode mahasiswa dari awal sampai akhir, manual."""
    banding = 0
    for i in range(len(data)):
        banding += 1  # satu perbandingan data[i][0] == target
        if data[i][0] == target:
            return i, banding
    return -1, banding

# Uji 1: target yang pasti ada
idx1, bd1 = linear_search_nim(data, target_nim)
print(f"Linear {target_nim}: indeks={idx1}, perbandingan={bd1}")

# Uji 2: kode yang tidak ada
idx2, bd2 = linear_search_nim(data, tidak_ada)
print(f"Linear {tidak_ada}: indeks={idx2}, perbandingan={bd2}")
```

## 1.3 Merge Sort rekursif (berdasarkan kode mahasiswa)
Pola modul Pertemuan 7: belah sampai 1 elemen, lalu `merge` dua bagian yang sudah urut. Data awal tidak diubah - sortir memakai salinan (`row[:]` per baris).

```python
def merge(left, right):
    """Gabung dua list yang masing-masing sudah urut (bandingkan kolom kode)."""
    hasil = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i][0] <= right[j][0]:
            hasil.append(left[i])
            i += 1
        else:
            hasil.append(right[j])
            j += 1
    # sisa salah satu sisi langsung ditempel (sudah urut)
    while i < len(left):
        hasil.append(left[i])
        i += 1
    while j < len(right):
        hasil.append(right[j])
        j += 1
    return hasil


def merge_sort_nim(data):
    """Urutkan berdasarkan kode mahasiswa, wajib rekursi."""
    if len(data) <= 1:
        return data
    tengah = len(data) // 2
    kiri = merge_sort_nim(data[:tengah])
    kanan = merge_sort_nim(data[tengah:])
    return merge(kiri, kanan)

# Salinan supaya data awal tidak ikut berubah
salinan = [baris[:] for baris in data]
data_urut = merge_sort_nim(salinan)

print("Sebelum sorting (data awal tetap):")
for i, row in enumerate(data):
    print(f"  [{i:2d}] {row[0]} {row[1]:6s} {row[2]}")
print("\nSesudah sorting:")
for i, row in enumerate(data_urut):
    print(f"  [{i:2d}] {row[0]} {row[1]:6s} {row[2]}")
```

## 1.3b Trace Merge Sort pada 6 data pertama (opsi a)
Dipilih opsi (a): proses Merge Sort pada 6 data pertama. Tiap baris `masuk` adalah satu panggilan rekursif yang membelah data, tiap baris `merge` adalah penggabungan dua bagian yang sudah urut. Makin ke kanan (indentasi) makin dalam rekursinya.


```python
# Trace Merge Sort 6 data pertama (opsi a) - gaya modul Pertemuan 7
def merge_sort_trace(data, depth=0):
    indent = '  ' * depth
    print(indent + 'masuk: ' + str([r[0] for r in data]))
    if len(data) <= 1:
        return data
    tengah = len(data) // 2
    kiri = merge_sort_trace(data[:tengah], depth + 1)
    kanan = merge_sort_trace(data[tengah:], depth + 1)
    hasil = merge(kiri, kanan)
    print(indent + 'merge: ' + str([r[0] for r in hasil]))
    return hasil

enam = [baris[:] for baris in data[:6]]
print('6 data pertama: ' + str([r[0] for r in enam]))
hasil6 = merge_sort_trace(enam)
print('hasil akhir 6 data: ' + str([r[0] for r in hasil6]))
```

## 1.4 Binary Search manual (data harus sudah urut)
Jepit dengan `low/high`, ambil tengah `mid`. Tiap cek `data_urut[mid]` dihitung 1 perbandingan.

```python
def binary_search_nim(data_urut, target):
    """Binary search manual pada data yang sudah urut."""
    low, high = 0, len(data_urut) - 1
    banding = 0
    while low <= high:
        mid = (low + high) // 2
        banding += 1  # satu perbandingan data_urut[mid][0] vs target
        if data_urut[mid][0] == target:
            return mid, banding
        elif data_urut[mid][0] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, banding

# Uji pada data yang sudah terurut
b_idx1, b_bd1 = binary_search_nim(data_urut, target_nim)
print(f"Binary {target_nim}: indeks={b_idx1}, perbandingan={b_bd1}")
b_idx2, b_bd2 = binary_search_nim(data_urut, tidak_ada)
print(f"Binary {tidak_ada}: indeks={b_idx2}, perbandingan={b_bd2}")
```

## 1.5 Trace / dry run Binary Search untuk target_nim
Dipilih opsi (b): Binary Search untuk target. Setiap langkah menampilkan batas `low/high/mid` sehingga terlihat ruang pencarian menyusut setengah.

```python
# Trace Binary Search untuk target_nim (opsi b)
print(f"Trace Binary Search, target = {target_nim}:")
low, high = 0, len(data_urut) - 1
langkah = 0
while low <= high:
    mid = (low + high) // 2
    langkah += 1
    kode = data_urut[mid][0]
    print(f"  langkah {langkah}: low={low} high={high} mid={mid} -> {kode}", end="")
    if kode == target_nim:
        print("  KETEMU")
        break
    elif kode < target_nim:
        print("  (< target, geser low ke mid+1)")
        low = mid + 1
    else:
        print("  (> target, geser high ke mid-1)")
        high = mid - 1
print(f"Selesai dalam {langkah} perbandingan (ruang menyusut setengah tiap langkah).")
```

## 1.7 Analisis kompleksitas (Soal 1)

**Linear Search - O(n) waktu, O(1) ruang.** Tiap elemen dicek satu kali; kasus terburuk seluruh n elemen dibandingkan (terbukti di atas: `TI999` butuh 12 perbandingan untuk n=12). Tidak butuh data urut, tidak butuh memori tambahan.

**Binary Search - O(log n) waktu, O(1) ruang (versi iteratif).** Tiap langkah membuang setengah ruang (`low/high` pada trace di atas), jadi untuk n=12 cukup 4 perbandingan. Syarat mutlak: **data harus terurut**, karena keputusan buang kiri/kanan hanya valid jika urutan monoton - pada data acak, separuh yang dibuang bisa saja berisi target.

**Merge Sort - O(n log n) waktu, O(n) ruang tambahan.** Membelah sedalam log n lalu menggabung (merge) yang totalnya O(n) per level. Butuh list hasil sementara sebesar n (plus potongan `data[:mid]` saat rekursi), jadi tidak in-place seperti Bubble/Insertion (O(1)).

**Kapan biaya sorting layak dibayar?** Sorting O(n log n) hanya impas jika pencarian dilakukan **berulang-ulang**: total = sort sekali + q x log n. Untuk 1-2 kali cari (seperti tugas ini), Linear O(n) justru lebih murah. Sorting baru menang saat q besar (mis. sistem akademik yang mencari ribuan kali per hari).
**Angka dari pengujian di atas.** Target TI005: Linear 5 vs Binary 4 perbandingan. TI999: Linear 12 vs Binary 4. Binary menang per pencarian, tapi ia menuntut bayar sorting dulu O(n log n). Untuk tugas ini (n=12, hanya 2 kali cari) total Linear = 5 + 12 = 17 perbandingan, sedangkan jalur sort + binary jauh lebih besar. Sorting baru impas kalau pencarian diulang ratusan kali (sistem akademik harian): bayar sort sekali, lalu tiap cari hanya O(log n).

# SOAL 2 - Antrean Layanan (Linked Queue) + Undo (Stack)

Queue = antrean aktif (FIFO), Stack = riwayat yang sudah dilayani (LIFO). Queue wajib singly linked list (`front/rear/count`), dilarang `deque`/`Queue`/`pop(0)` sebagai implementasi utama. Stack dibungkus class sendiri di atas Python list.

## 2.1 Class Node, LinkedQueue, Stack + uji operasi dasar
Uji dasar dulu sebelum simulasi utama, sesuai urutan tugas. `dequeue`/`pop` kosong ditangani aman (cetak pesan, kembalikan `None`, program tidak berhenti).

```python
class Node:
    """Satu simpul singly linked list."""
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedQueue:
    """Queue FIFO di atas singly linked list (front = keluar, rear = masuk)."""
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def is_empty(self):
        return self.front is None

    def size(self):
        return self.count

    def enqueue(self, data):  # masuk dari belakang
        baru = Node(data)
        if self.is_empty():
            self.front = baru
            self.rear = baru
        else:
            self.rear.next = baru
            self.rear = baru
        self.count += 1

    def enqueue_front(self, data):  # untuk UNDO: kembalikan ke depan
        baru = Node(data)
        baru.next = self.front
        self.front = baru
        if self.rear is None:  # queue tadinya kosong
            self.rear = baru
        self.count += 1

    def dequeue(self):  # keluar dari depan
        if self.is_empty():
            print("Queue kosong: tidak ada yang bisa dilayani.")
            return None
        keluar = self.front.data
        self.front = self.front.next
        if self.front is None:  # jadi kosong
            self.rear = None
        self.count -= 1
        return keluar

    def peek(self):
        if self.is_empty():
            print("Queue kosong: tidak ada front.")
            return None
        return self.front.data

    def display(self):
        isi, cur = [], self.front
        while cur is not None:
            isi.append(cur.data)
            cur = cur.next
        return isi


class Stack:
    """Riwayat LIFO; penyimpanan internal boleh Python list."""
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            print("Stack kosong: tidak ada riwayat untuk di-undo.")
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            print("Stack kosong: tidak ada top.")
            return None
        return self.items[-1]

    def display(self):
        return list(self.items)

# --- Uji dasar sebelum simulasi utama ---
q0, s0 = LinkedQueue(), Stack()
q0.enqueue("X1")
q0.enqueue("X2")
q0.enqueue_front("X0")
print("Uji Queue :", q0.display(), "| peek =", q0.peek(), "| size =", q0.size())
print("Uji dequeue:", q0.dequeue(), "-> sisa:", q0.display())
s0.push("R1")
s0.push("R2")
print("Uji Stack :", s0.display(), "| peek =", s0.peek(), "| size =", s0.size())
print("Uji pop:", s0.pop(), "-> sisa:", s0.display())
```

## 2.2 Simulasi 12 operasi
d=4 -> A=MHS-401 ... F=MHS-406. LAYANI = `dequeue` lalu `push` ke riwayat; UNDO = `pop` riwayat lalu `enqueue_front`. Setelah tiap langkah tampilkan Queue (front->rear) dan Stack.

```python
# ID layanan dari digit terakhir NIM (d = 4)
A = f'MHS-{d}01'
B = f'MHS-{d}02'
C = f'MHS-{d}03'
D = f'MHS-{d}04'
E = f'MHS-{d}05'
F = f'MHS-{d}06'
print("ID:", A, B, C, D, E, F)

queue = LinkedQueue()
riwayat = Stack()


def status(langkah, operasi):
    print(f"[{langkah:2d}] {operasi:10s} | Queue(front->rear): {queue.display()} | Stack(riwayat): {riwayat.display()}")


def layani():
    dilayani = queue.dequeue()  # gagal -> None, jangan push
    if dilayani is not None:
        riwayat.push(dilayani)
    return dilayani


def undo():
    batal = riwayat.pop()  # gagal -> None, jangan enqueue
    if batal is not None:
        queue.enqueue_front(batal)
    return batal

# 12 operasi sesuai tabel soal, berurutan
queue.enqueue(A);            status(1, "DATANG A")
queue.enqueue(B);            status(2, "DATANG B")
queue.enqueue(C);            status(3, "DATANG C")
layani();                    status(4, "LAYANI")
queue.enqueue(D);            status(5, "DATANG D")
layani();                    status(6, "LAYANI")
queue.enqueue(E);            status(7, "DATANG E")
undo();                      status(8, "UNDO")
layani();                    status(9, "LAYANI")
layani();                    status(10, "LAYANI")
undo();                      status(11, "UNDO")
queue.enqueue(F);            status(12, "DATANG F")
```

## 2.4 Analisis (Soal 2)

**FIFO vs LIFO dari simulasi.** Queue FIFO: yang datang duluan dilayani duluan - langkah 4 melayani A (bukan B/C), langkah 6 melayani B; C yang datang ketiga baru keluar di langkah 9. Stack LIFO: riwayat `[A, B]` - UNDO langkah 8 mengambil B (terakhir masuk, keluar duluan) dan mengembalikannya ke depan antrean; UNDO langkah 11 mengambil C (paling atas saat itu), bukan A yang paling bawah.

**Kompleksitas LinkedQueue:** `enqueue`, `enqueue_front`, `dequeue`, `peek` semuanya **O(1)**. Alasannya: `front` dan `rear` adalah referensi langsung ke ujung - tidak ada traversal. `enqueue` hanya `rear.next = baru`, `dequeue` hanya `front = front.next`, `enqueue_front` hanya menyambung satu node di depan. `display`/`size` beda: `display` O(n) karena menelusuri semua node, `size` O(1) karena memakai `count`.

**Kompleksitas Stack (list internal):** `push`, `pop`, `peek` semuanya **O(1)** rata-rata (operasi di ujung list, `append`/`pop()` tanpa geser elemen).

**Mengapa linked list, bukan `list.pop(0)`?** `pop(0)` pada Python list menggeser **seluruh** sisa elemen satu posisi -> tiap `dequeue` berharga **O(n)**. Untuk loket yang melayani ratusan antrean, totalnya O(n^2). Linked list hanya memindah pointer `front` -> O(1) per `dequeue`, tanpa geser apa pun, dan ukurannya tumbuh dinamis tanpa realokasi array.
